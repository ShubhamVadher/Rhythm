```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/spotify-dataset.csv")
df.head()
```

```python
df.info()
```

```python
print("Songs:", len(df))
print("Unique Songs:", df["track_id"].nunique())
print("Artists:", df["artists"].nunique())
print("Genres:", df["track_genre"].nunique())
```

```python
df["track_id"].duplicated().sum()
```

```python
genre_counts = (
    df.groupby("track_id")["track_genre"]
      .nunique()
      .sort_values(ascending=False)
)

genre_counts.head(20)
```

```python
result = df.groupby("track_id")["track_genre"].nunique().value_counts().reset_index()
result.columns = ["num_genres" , "count"]
result
```

```python
song_df = df.drop_duplicates("track_id").drop(columns = "track_genre")
```

```python
genre_df = df.groupby("track_id")["track_genre"].unique().reset_index()
song_df = pd.merge(song_df,genre_df,on = "track_id")
```

```python
song_df.drop(columns="Unnamed: 0" , inplace = True)
```

```python
song_df.shape
```

```python
song_df.to_csv(
    "../data/processed/spotify-dataset_clean.csv",
    index=False
)
```
